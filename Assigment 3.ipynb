```python
num = input("Enter Table Number : ")
if num.isdigit():
    num = int(num)

    for i in range(1,10+1):
        print(f"{num} x {i} = {num*i}")
else:
    print("Invalid Command.")
        
```

```python
Pass = 0
Fail = 0

for i in range(5):
    marks = input("Enter Marks : ")

    if marks.isdigit():
        marks = int(marks)

        if marks >= 50:
            Pass += 1
        else:
            Fail += 1    

    else:
        print("Invalid Command")    

print(f"Total Passed : {Pass}")
print(f"Total Failed : {Fail}")
```
