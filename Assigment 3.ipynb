```python
balance = 50000
while True:
    print("-------------------------")
    print("|  Bank Account Status  |")
    print("-------------------------\n")

    print("1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit\n")

    user = input("Enter Command (1-4) : ")
    print()

    #Check Balance
    if user == "1":
        print(f"Current Balance : {balance}\n")

    #Deposit
    elif user == "2":
        amount = input("Enter Amount : ")
        if amount.isdigit():
            amount = int(amount)
            balance = amount + balance
            print("-------------------------")
            print("|  Deposit Sucessfully  |")
            print("-------------------------\n")

            print(f"New Balance : {balance}\n")
        else:
            print("Invalid Command.\n")     

    #Withdraw        
    elif user == "3":
        amount = input("Enter Amount : ")
        if amount.isdigit():
            amount = int(amount)
            if amount <= balance:
                balance =  balance - amount
                print("---------------------------------")
                print("|  Withdraw Amount Sucessfully  |")
                print("---------------------------------\n")

                print(f"New Balance : {balance}\n")
            else:    
                print("Insufficient Balance.\n")
        else:
            print("Invalid Command.\n")   

    #Exit
    elif user == "4":
        break
    else:
        print("Inavlid Command.\n")             

    
```

```python
secret = 25
attempt = 0

while True:
    guess = input("Enter Guess : ")
    attempt += 1

    if guess.isdigit():
        guess = int(guess)

        if guess == secret:
            print(f"Attemt : {attempt}")
            print(f"{guess} : Your Guess is Correct.")
            break
        elif guess > secret:
            print(f"Attemt : {attempt}")
            print(f"{guess} : Too High\n")
        else:
            print(f"Attemt : {attempt}")
            print(f"{guess} : Too Low\n")        
    else:
        print("Invalid Command.\n")    
```

```python
total_student = 0
passed_student = 0
failed_student = 0

while True:
    marks = input("Enter Marks : ")
    if marks.isdigit() or marks.startswith('-'):
        marks = int(marks)
        if marks == -1:
            break
        elif marks >= 50:
            passed_student += 1
            total_student += 1
        elif marks < 50:   
            total_student += 1
            failed_student += 1 
    else:
        print("Invalid Command.")    

print(f"Total Student : {total_student}")
print(f"Passed Student : {passed_student}")
print(f"Failed Student : {failed_student}")        

```

```python

print("-------------------------------------------")
print("|  Student Performance Management System  |")
print("-------------------------------------------\n")

result = input("Enter Total Students : ")

if result.isdigit() and int(result) > 0:
    result = int(result)

    passed = 0
    failed = 0
    class_total = 0
    highest_percentage = 0
    lowest_percentage = 100

    for i in range(result):
        print("\nStudent No:", i + 1)
        print("---------------------------")

        Student_name = input("Enter Student Name : ")
        python = input("Marks in Python : ")
        maths = input("Marks in Mathematics : ")
        english = input("Marks in English : ")

        if (python.isdigit() and maths.isdigit()
                and english.isdigit()):

            python = int(python)
            maths = int(maths)
            english = int(english)

            Total_marks = 300

            if (python <= 100 and maths <= 100
                    and english <= 100):

                obtain_marks = python + maths + english
                percentage = obtain_marks / Total_marks * 100

                class_total += obtain_marks

                print("\nStudent Name :", Student_name)
                print("Obtained Marks :", obtain_marks, "/ 300")
                print("Percentage :", round(percentage, 2), "%")

                if (percentage >= 50 and python >= 40
                        and maths >= 40 and english >= 40):

                    passed += 1
                    print("Result : Pass")

                    if percentage >= 90:
                        print("Grade : A+")

                    elif percentage >= 80:
                        print("Grade : A")

                    elif percentage >= 70:
                        print("Grade : B")

                    elif percentage >= 60:
                        print("Grade : C")

                    else:
                        print("Grade : D")

                else:
                    failed += 1
                    print("Result : Fail")
                    print("Grade : Not Assigned")

                if percentage > highest_percentage:
                    highest_percentage = percentage

                if percentage < lowest_percentage:
                    lowest_percentage = percentage

            else:
                print("Invalid Marks! Marks must be between 0 and 100.")
                print("Please run the program again for this student.")

        else:
            print("Invalid Command! Enter numeric marks only.")
            print("Please run the program again for this student.")

    print("\n===========================================")
    print("           CLASS FINAL REPORT")
    print("===========================================")

    print("Total Students :", result)
    print("Passed Students :", passed)
    print("Failed Students :", failed)

    average_percentage = (class_total / (result * 300)) * 100

    print("Class Average Percentage :",
          round(average_percentage, 2), "%")

    print("Highest Percentage :",
          round(highest_percentage, 2), "%")

    print("Lowest Percentage :",
          round(lowest_percentage, 2), "%")

    print("===========================================")

else:
    print("Invalid Command! Enter a valid number of students.")
```

```python

print("-------------------------------------")
print("|  Restaurant Ordering & Billing    |")
print("-------------------------------------\n")

total_bill = 0
total_items = 0

while True:
    print("\n------------- MENU -------------")
    print("1. Burger       Rs. 500")
    print("2. Pizza        Rs. 1000")
    print("3. Biryani      Rs. 350")
    print("4. Fries        Rs. 250")
    print("5. Drink        Rs. 150")
    print("6. Checkout")
    print("--------------------------------")

    choice = input("Enter Your Choice : ")

    if choice.isdigit():
        choice = int(choice)

        if choice == 1:
            total_bill += 500
            total_items += 1
            print("Burger added.")

        elif choice == 2:
            total_bill += 1000
            total_items += 1
            print("Pizza added.")

        elif choice == 3:
            total_bill += 350
            total_items += 1
            print("Biryani added.")

        elif choice == 4:
            total_bill += 250
            total_items += 1
            print("Fries added.")

        elif choice == 5:
            total_bill += 150
            total_items += 1
            print("Drink added.")

        elif choice == 6:
            break

        else:
            print("Invalid choice. Please try again.")

    else:
        print("Invalid choice. Please try again.")


# Discount Calculation

discount = 0

if total_bill >= 5000:
    discount = total_bill * 20 / 100

elif total_bill >= 3000:
    discount = total_bill * 15 / 100

elif total_bill >= 1500:
    discount = total_bill * 10 / 100

discounted_bill = total_bill - discount


# Delivery Charges

if discounted_bill >= 3000:
    delivery_charges = 0
    print("\nFree Delivery!")

else:
    delivery_charges = 200


# Final Bill

final_bill = discounted_bill + delivery_charges

print("\n================================")
print("          FINAL BILL")
print("================================")

print("Total Items :", total_items)
print("Original Bill : Rs.", total_bill)
print("Discount : Rs.", discount)
print("Delivery Charges : Rs.", delivery_charges)
print("Final Bill : Rs.", final_bill)

print("\nThank you for ordering!")
```
